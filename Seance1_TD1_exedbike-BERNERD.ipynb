**Question 3** - Donnez la solution optimale, identifiez les ressources saturées, et interprétez économiquement.

**Réponse**

L'optimum est atteint pour la production de 75 modèles URBAN et 30 modèles CARGO. La marge max est donc de 15000 €. 

Les ressources "main-d'oeuvre" et "aluminimum" sont saturées. Ce sont elles qui limitent l'augmentation de la marge maximale atteignable. Il reste en revanche 30 h de machine inutilisée (s3). Cela ne servirait à rien de rajouter du temps machine supplémentaire.

Les prix duaux sont de 25 €/h pour la main d'oeuvre et de 10/3 €/kg pour l'aluminium. Cela signifie qu'une heure de main d'oeuvre supplémentaire pourrait augmenter la marge de 25 euros, et 1 kg d'aluminimium supplémentaire pourrait augmenter la marge de 3,33 euros. Ces prix duaux sont à comparer avec le coût réel d'1h de main d'oeuvre supplémentaire ou d'achat d'1kg d'aluminium supplémentaire pour savoir si ce serait rentable d'augmenter ces 2 ressources pour augmenter la marge.


**Question 4** - En quoi cette résolution à la main diffère-t-elle de celle du programme complet à trois produits de la Partie projet ? Pourquoi l’entreprise a-t-elle malgré tout intérêt à utiliser un solveur pour ce dernier plutôt que de reproduire cette méthode à la main ?

**Réponse**

Avec 2 variables, la méthode manuelle reste accessible, le polytope est un polygone 2D, donc facilement interprétable et vérifiable visuellement.
Avec 3 variables, le polytope devient 3D, plus complexe à visualiser et interpréter.
Un solveur sera plus rapide, plus fiable qu'à la main. Il permet également de "jouer" plus facilement avec les paramètres pour étudier et vérifier plusieurs variantes, ou de mettre à jour plus rapidement et simplement les modèles (évolution modèles de coût par ex)
